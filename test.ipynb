```python
from qiskit import QuantumCircuit
from qiskit_ibm_runtime.fake_provider import FakeManilaV2
from qviz import compile_circuit, display, interactive

circuit = QuantumCircuit(3)
circuit.h(0)
circuit.cx(0, 1)
circuit.cx(0, 1)
circuit.cz(0, 1)
circuit.cx(1, 0)
circuit.cx(0, 1)
circuit.cx(0, 1)
circuit.cx(1, 2)
circuit.measure_all()

backend = FakeManilaV2()

# JSON-serializable visualization model
model = compile_circuit(circuit, backend)

# Interactive visualization in Jupyter
display(circuit, backend)
```
